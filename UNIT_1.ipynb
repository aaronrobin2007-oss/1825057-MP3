Unit-1: Handling of Complex Numbers

```python
#w=z^2

import numpy as np
import matplotlib.pyplot as plt
z=np.array([1-5j,1-4j,1-3j,1-3j,1-1j,1+0j,1+1j,1+2j,1+3j,1+4j,1+5j])
fz=z**2
plt.subplot(1,2,1)
plt.plot(z.real,z.imag,color='blue',marker='o')
plt.title('z plane')
plt.subplot(1,2,2)
plt.plot(fz.real,fz.imag,color='red',marker='o')
plt.title('w plane')
plt.show()
```

```python
#2D rotation of complex numbers

import numpy as np
import matplotlib.pyplot as plt
import math as mt
theta=mt.radians(5)
r=np.array([[np.cos(theta),-np.sin(theta)],
            [np.sin(theta),np.cos(theta)]])
a=np.array([5,5])
plt.scatter(a[0],a[1],color='black')
for i in range(72):
  a=r@a
  plt.scatter(a[0],a[1],color='blue')
plt.title('2D rotation')
plt.show()

```

```python
#nth root of complex number in cartesian and polar form

import numpy as np
import matplotlib.pyplot as plt
z = -8 + 11j
n = 3
r = np.abs(z)
theta = np.angle(z)
r_root = r ** (1 / n)
roots = []
for k in range(n):
    theta_k = (theta + 2 * np.pi * k) / n
    z_k = r_root * np.exp(1j * theta_k)
    roots.append(z_k)
    print("Root", k, "-> Polar:", np.round(r_root, 2), np.round(theta_k, 2), "| Cartesian:", np.round(z_k, 2))
roots = np.array(roots)
plt.scatter(roots.real, roots.imag, color="blue")
plt.axhline(0, color="black")
plt.axvline(0, color="black")
plt.title("nth Roots of a Complex Number")
plt.show()
```
