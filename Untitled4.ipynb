```python
import numpy as np
import matplotlib.pyplot as plt
z=np.array([1-5j,1-4j,1-3j,1-2j,1-1j,1-0j,1+1j,1+2j])
fz=z**2
plt.subplot(1,2,1)
plt.title('f(z)=z')
plt.plot(z.real,z.imag,color='red',marker='o')
plt.subplot(1,2,2)
plt.title('f(z)=z^2')
plt.plot(fz.real,fz.imag,color='blue',marker='o')
plt.show()
```
